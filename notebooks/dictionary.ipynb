# The classical ↔ quantum dictionary

This is the one table that gathers what the course translates between the classical world and the quantum one. Each row reads from left to right: a classical object on the left, and the quantum object that plays its part on the right. Some rows are translations, some are contrasts, and a contrast is as much a result as a match.

The last column, **Earned in**, names the notebook or notebooks where you build that row yourself, written as `02/02` for module 02, notebook 02. Each of those notebooks shows the rows it earns in a short table of its own and links back here, so you can open the full picture whenever you wish and watch it grow as you go.

## Module 01 — Quantum Foundations

| Classical | Quantum | Earned in |
|---|---|---|
| bit ($0$ or $1$) | qubit $a_0\,|0\rangle + a_1\,|1\rangle$ (two complex amplitudes) | `01/01` |
| probabilities only add (non-negative) | amplitudes add and cancel: interference | `01/02` |
| a bit's state space: two isolated points | a pure qubit's state space: the Bloch sphere | `01/03` |
| outcome probabilities $p_x$ | Born rule $p_x = \lvert a_x\rvert^2$ | `01/04` |
| overlap of two probability vectors, $\sum_x \sqrt{p_x q_x}$ (Bhattacharyya coefficient) | inner product $\langle\varphi|\psi\rangle$ | `01/05` |
| a reversible operation on a bit: a permutation (identity, NOT) | a gate: a unitary matrix $U$, $U^\dagger U = I$ | `01/06` |
| a quantity $f(x)$ of the bit, average $\sum_x p_x f(x)$ | an observable: a Hermitian matrix $A = A^\dagger$, average $\langle\psi|A|\psi\rangle$ | `01/07` |
| one way to read a bit: look at it | measurement in any orthonormal basis $\{|b_0\rangle,|b_1\rangle\}$, $P(k) = |\langle b_k|\psi\rangle|^2$ | `01/08` |
| a bit has no continuous knob: you flip it or you do not | rotation gates $R_x, R_y, R_z$ and phase gates $P(\lambda)$: continuous dials | `01/09` |
| average $\sum_x p_x f(x)$, estimated from $N$ samples with standard error $\sqrt{\mathrm{Var}(f)/N}$ | expectation value $\langle A\rangle = \sum_k \lambda_k P(k)$, estimated from $N$ shots of a $\pm1$ observable with standard error $\sqrt{(1 - \langle A\rangle^2)/N}$ | `01/10` |
| a randomized program on bits, run many times to estimate its output frequencies | a circuit run for $N$ shots, whose counts sample the Born shadow | `01/11` |
| probability vector $p = (p_0, p_1)$ | density matrix $\rho = |\psi\rangle\langle\psi|$ (pure case); a diagonal $\rho$ is a classical $p$ | `01/12` |
| a coin $p$: not knowing which bit you hold, the only kind of uncertainty | mixed state $\rho = \sum_i p_i|\psi_i\rangle\langle\psi_i|$: ignorance of which state, a second kind, distinct from superposition | `01/13` |
| total variation $\tfrac12\sum_x|p_x - q_x|$ and Bhattacharyya overlap $\sum_x\sqrt{p_x q_x}$ | trace distance $T = \tfrac12\|\rho - \sigma\|_1$ and fidelity $F = (\mathrm{tr}\sqrt{\sqrt\rho\sigma\sqrt\rho})^2$ | `01/14` |
| estimating a probability vector from counted frequencies | state tomography: $\hat\rho = \tfrac12(I + \hat{\vec r}\cdot\vec\sigma)$ from the $X$, $Y$, $Z$ counts of three circuits | `01/15` |
| marginal of a joint table, $p_A(a) = \sum_b p(a,b)$ | partial trace $\rho_A = \mathrm{tr}_B\,\rho_{AB}$ | `01/16` |
| product distribution $p_A(a)\,p_B(b)$ | product state $\rho_A \otimes \rho_B$ | `01/16` |
| XOR on two bits, $(a, b) \mapsto (a, a \oplus b)$ | CNOT $= |0\rangle\langle0|\otimes I + |1\rangle\langle1|\otimes X$ | `01/17` |
| a certain joint table (one outcome, probability $1$) has certain marginals | entanglement: a pure non-product state ($\det\Psi \ne 0$) has mixed parts | `01/18` |
| stochastic matrix (Markov kernel) $p \to Tp$ | quantum channel $\rho \to \sum_k K_k\rho K_k^\dagger$, $\sum_k K_k^\dagger K_k = I$ | `01/19` |

## Module 02 — Information and Geometry

| Classical | Quantum | Earned in |
|---|---|---|
| Shannon entropy $H(p) = -\sum_x p_x\log_2 p_x$ | von Neumann entropy $S(\rho) = -\mathrm{tr}\,\rho\log_2\rho$ | `02/01` |
| relative entropy (Kullback–Leibler) $D(p\|q) = \sum_x p_x\log_2\frac{p_x}{q_x}$ | Umegaki relative entropy $S(\rho\|\sigma) = \mathrm{tr}\,\rho(\log_2\rho - \log_2\sigma)$ (`02/09`) | `02/02`, `02/12` |
| mutual information $I(X;Y) = D(p_{XY}\,\|\,p_X \otimes p_Y) \le \min(H(X), H(Y))$ | quantum mutual information $I(A{:}B) = S(A) + S(B) - S(AB)$, up to twice the classical bound (`02/10`) | `02/03`, `02/12` |
| transfer entropy $\mathrm{TE}_{Y\to X} = I(X_{t+1};\,Y_t\,|\,X_t)$ | no quantum lift in this course: the quantum coupling measures of module 04 are symmetric | `02/04` |
