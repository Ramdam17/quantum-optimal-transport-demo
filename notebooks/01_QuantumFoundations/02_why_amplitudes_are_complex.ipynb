# 02 — Why Amplitudes Are Complex

Answer the question `01/01` set aside: why a qubit's amplitudes are *complex* numbers, and what the phase they carry actually does.

**Prerequisites:** `01/01_amplitudes_and_superposition` (a qubit as two amplitudes, the modulus $|a|=\sqrt{x^2+y^2}$, normalisation).

**What you'll be able to do**
- Write an amplitude in polar form $a = r\,e^{i\varphi}$ (Euler's formula) and read off its magnitude and phase.
- Run a by-hand *recombination rule* on two amplitudes and watch them reinforce or cancel — interference — and see why a secretly-landed coin cannot do this.
- Explain why real, signed amplitudes are not enough: exhibit a state that only complex amplitudes can hold.
- Distinguish a *global* phase (invisible) from a *relative* phase (physical).

In `01/01` you built a qubit from two amplitudes and met a tilted state whose amplitude on $|1\rangle$ was complex. We waved at the phase and moved on. Today we earn it, in two steps. First, we show that amplitudes must be able to *cancel* — which already rules out non-negative weights and delivers the promise `01/01` made: a superposition is not a coin that secretly landed. Then we go one step further and show that *signed real* numbers are still not enough — there is a perfectly good qubit state that no real pair of amplitudes can hold. That is what makes $\mathbb{C}$ the right home.

```python
import numpy as np
import matplotlib.pyplot as plt

from qot_course import viz
from qot_course.colors import COLORS
from qot_course.quantum.states import KET_0, KET_1, KET_PLUS, KET_MINUS, qubit_state

viz.use_course_style()
```

## 1. Two states that weights cannot tell apart

Here are two superpositions. Each puts the *same* amount of weight on $|0\rangle$ and on $|1\rangle$:

$$ |+\rangle = \tfrac{1}{\sqrt2}\big(|0\rangle + |1\rangle\big), \qquad |-\rangle = \tfrac{1}{\sqrt2}\big(|0\rangle - |1\rangle\big). $$

The only difference is a minus sign on the second amplitude. Let's look at the amplitudes and at their squared sizes (the *weights* $|a|^2$ of `01/01`).

```python
for name, state in [("|+>", KET_PLUS), ("|->", KET_MINUS)]:
    print(f"{name}: amplitudes = {np.round(state, 3)}   weights |a|^2 = {np.round(np.abs(state)**2, 3)}")
```

**Read the output.** Both states carry weight $0.5$ on each outcome — identical, as far as a list of weights is concerned. A description by weights alone ("50% this, 50% that") *cannot* tell $|+\rangle$ from $|-\rangle$. Are they nevertheless different? The sign is the simplest possible phase, and we are about to build the test that makes it visible. First we need a way to name a phase in general.

## 2. Polar form: magnitude and phase

A complex number is a point in the plane, and a point in the plane can be given by a length and an angle. **Euler's formula** is the bridge:

$$ e^{i\varphi} = \cos\varphi + i\sin\varphi , $$

a number of modulus $1$ (check: $\cos^2\varphi+\sin^2\varphi=1$) sitting at angle $\varphi$ on the unit circle. So every amplitude splits cleanly into a size and an angle,

$$ a = r\,e^{i\varphi}, \qquad r = |a| \ge 0, \quad \varphi = \arg(a) \in (-\pi, \pi]. $$

The magnitude $r$ sets the *weight* $r^2$ (in `01/04` this becomes the probability of the outcome — the Born rule). The angle $\varphi$ is the new degree of freedom. In NumPy, `np.abs` gives $r$ and `np.angle` gives $\arg(a)$ in $(-\pi,\pi]$. Let's take the exact tilted amplitude from `01/01`.

```python
a = qubit_state(theta=2 * np.pi / 3, phi=np.pi / 4)[1]   # the |1> amplitude of the tilted state from 01/01
r, phi_a = np.abs(a), np.angle(a)
print(f"a = {np.round(a, 4)}")
print(f"magnitude r = {r:.4f}   ->   weight r^2 = {r**2:.4f}")
print(f"phase phi   = {phi_a:.4f} rad = {np.degrees(phi_a):.1f} deg")
print(f"reconstructed r*e^(i*phi) = {np.round(r * np.exp(1j * phi_a), 4)}")
print(f"Euler check: e^(i*phi) = {np.round(np.exp(1j * phi_a), 4)}  vs  cos+i sin = {np.round(np.cos(phi_a) + 1j*np.sin(phi_a), 4)}")
```

**Read the output.** The single complex number $a$ decomposes into a magnitude ($\sin(\pi/3)\approx 0.866$, so weight $0.75$) and a phase ($45^\circ$). Rebuilding $r\,e^{i\varphi}$ returns the original amplitude to machine precision, and Euler's formula checks out numerically. Let's picture this amplitude as an arrow in the complex plane.

```python
fig, ax = plt.subplots(figsize=(5.2, 5.2))
circle = plt.Circle((0, 0), 1.0, fill=False, color=COLORS["grid"], lw=1.5)
ax.add_artist(circle)
# the tilted amplitude a = r e^{i phi}: length r, angle phi
ax.annotate("", xy=(a.real, a.imag), xytext=(0, 0),
            arrowprops=dict(arrowstyle="->", color=COLORS["quantum"], lw=2.5))
ax.text(1.12 * a.real, 1.12 * a.imag, r"$a = r\,e^{i\varphi}$", color=COLORS["quantum"], ha="left")
# the |1> amplitudes of |+> and |->: same length, angles 0 and pi
for s, col, lab in [(KET_PLUS[1], COLORS["source"], r"$|+\rangle$: $\varphi=0$"),
                    (KET_MINUS[1], COLORS["target"], r"$|-\rangle$: $\varphi=\pi$")]:
    ax.annotate("", xy=(s.real, s.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="->", color=col, lw=2.0))
    ax.text(s.real, s.imag - 0.16, lab, color=col, ha="center")
ax.axhline(0, color=COLORS["grid"], lw=1)
ax.axvline(0, color=COLORS["grid"], lw=1)
ax.set_xlim(-1.3, 1.3); ax.set_ylim(-1.3, 1.3)
ax.set_aspect("equal")
ax.set_xlabel("real part"); ax.set_ylabel("imaginary part")
ax.set_title("An amplitude is an arrow: length = magnitude, angle = phase", pad=12)
plt.show()
```

**Read the figure.** The tilted amplitude (sky blue) is an arrow of length $0.866$ at $45^\circ$; the unit circle is the reference for magnitude $1$. The $|1\rangle$ amplitudes of $|+\rangle$ (periwinkle) and $|-\rangle$ (amber) have the *same* length $0.707$ and point in opposite directions: phase $0$ versus phase $\pi$. Weights only see the length; the direction is exactly what section 1 could not detect. Now we build the test that does.

## 3. Interference: a recombination rule that makes the sign visible

Weights alone missed the sign. But amplitudes are not only *read* — they can be *recombined*. Here is the simplest rule that mixes the two amplitudes of a qubit, using nothing but addition and subtraction:

$$ (a_0,\ a_1)\ \longmapsto\ \Big(\tfrac{a_0 + a_1}{\sqrt2},\ \tfrac{a_0 - a_1}{\sqrt2}\Big). $$

Read it as: the two amplitudes both feed outcome $0$ (they *add*), and both feed outcome $1$ (they *subtract*). The $1/\sqrt2$ keeps the result normalised (check it: $|a_0+a_1|^2 + |a_0-a_1|^2 = 2(|a_0|^2+|a_1|^2)$). This is not a made-up trick — it is a real physical operation on a qubit, one of the most common gates on any quantum computer; in `01/08` it gets its name (Hadamard) and its full treatment. Today we only need its arithmetic. Apply it to $|+\rangle$ and $|-\rangle$.

```python
def recombine(state):
    """By-hand recombination rule: (a0, a1) -> ((a0 + a1)/sqrt2, (a0 - a1)/sqrt2)."""
    a0, a1 = state
    return np.array([a0 + a1, a0 - a1]) / np.sqrt(2)

for name, state in [("|+>", KET_PLUS), ("|->", KET_MINUS)]:
    out = recombine(state)
    print(f"{name}: recombined amplitudes = {np.round(out, 3)}   weights = {np.round(np.abs(out)**2, 3)}")
```

**Read the output.** The two states that were indistinguishable by weight now give *opposite, certain* results: $|+\rangle$ recombines to weights $(1, 0)$ — the two amplitudes reinforce on outcome $0$ and cancel on outcome $1$ — while $|-\rangle$ recombines to $(0, 1)$. That cancellation is **destructive interference**, and it is only possible because amplitudes can be negative: non-negative weights can pile up, never subtract.

This also settles the promise from `01/01`. A coin that had *secretly* landed on $0$ or on $1$ is the state $|0\rangle$ or $|1\rangle$; feed either through the same rule and you get weights $(0.5, 0.5)$ — a coin flip, never a certainty. So the equal-bars state $|+\rangle$ is not a hidden coin: run the recombination and it answers $0$ *every single time*. Let's check the coin claim, then turn the sign into a continuous dial.

```python
for name, state in [("|0> (coin landed 0)", KET_0), ("|1> (coin landed 1)", KET_1)]:
    out = recombine(state)
    print(f"{name}: recombined weights = {np.round(np.abs(out)**2, 3)}")
```

**Read the output.** Both "landed" coins recombine to $(0.5, 0.5)$; so would a coin that lands $0$ half the time and $1$ half the time. No secretly-landed coin reproduces the certainty that $|+\rangle$ shows. This is the first genuinely quantum fact of the course, and you computed it with two additions.

Now generalise the sign to any phase. Take $|\psi_\varphi\rangle = \tfrac{1}{\sqrt2}\big(|0\rangle + e^{i\varphi}|1\rangle\big)$: weights $(0.5, 0.5)$ for *every* $\varphi$, so weights see nothing. Recombine it, and the outcome-$0$ amplitude is

$$ s(\varphi) = \tfrac{1}{2}\big(1 + e^{i\varphi}\big). $$

By hand, using Euler: $|1+e^{i\varphi}|^2 = (1+\cos\varphi)^2 + \sin^2\varphi = 2 + 2\cos\varphi = 4\cos^2(\varphi/2)$, so the outcome-$0$ weight is $|s(\varphi)|^2 = \cos^2(\varphi/2)$ and the outcome-$1$ weight is $\sin^2(\varphi/2)$. Let's confirm numerically and draw the dial.

```python
phi = np.linspace(0, 2 * np.pi, 400)
psi_phi = np.array([np.ones_like(phi), np.exp(1j * phi)]) / np.sqrt(2)     # shape (2, 400)
out0, out1 = recombine(psi_phi)
weight0 = np.abs(out0) ** 2
print("numeric weight_0 == cos^2(phi/2) on the whole grid:", np.allclose(weight0, np.cos(phi / 2) ** 2))
print("weights sum to 1 on the whole grid              :", np.allclose(weight0 + np.abs(out1) ** 2, 1.0))

fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.plot(phi, weight0, color=COLORS["quantum"], lw=2.5, label=r"weight of outcome 0 after recombination")
ax.plot(phi, np.abs(out1) ** 2, color=COLORS["target"], lw=2.0, ls="--", label=r"weight of outcome 1")
ax.axhline(0, color=COLORS["grid"], lw=1)
ax.scatter([0, np.pi], [1, 0], color=COLORS["highlight"], zorder=5, s=60)
ax.annotate(r"$|+\rangle$: constructive", (0, 1), textcoords="offset points", xytext=(10, -14), color=COLORS["text"])
ax.annotate(r"$|-\rangle$: destructive", (np.pi, 0), textcoords="offset points", xytext=(10, 12), color=COLORS["text"])
ax.set_xlabel(r"relative phase $\varphi$ of $|\psi_\varphi\rangle$")
ax.set_ylabel("weight after recombination")
ax.set_title("Interference: a relative phase steers the outcome", pad=12)
ax.legend(loc="center right")
plt.show()
```

**Read the figure.** Every $|\psi_\varphi\rangle$ on this axis has weights $0.5/0.5$ *before* recombination — the input looks the same for all $\varphi$. After recombination the outcome-$0$ weight sweeps from $1$ ($\varphi=0$, that is $|+\rangle$) down to $0$ ($\varphi=\pi$, that is $|-\rangle$) and back, and the outcome-$1$ weight does the opposite; the two always sum to $1$. Nothing about the *sizes* of the input amplitudes changed — only their phase relationship. This dial is the mechanism quantum algorithms use to concentrate weight on the answer they want (Deutsch–Jozsa, Grover, the quantum Fourier transform all steer it — Nielsen & Chuang, §1.4 and ch. 5–6).

## 4. Signs are not enough: why *complex*

Be honest about what section 3 showed. Everything there used the two phases $\varphi=0$ and $\varphi=\pi$ — a plus and a minus sign — and the recombination rule itself is real. **Signed real amplitudes already interfere.** So the title question is still open: why does a qubit need the whole complex plane, and not just $\pm$?

Here is the answer, at the level of a single qubit. With real amplitudes, a normalised state is $\cos\vartheta\,|0\rangle + \sin\vartheta\,|1\rangle$ for some angle $\vartheta$ (two real numbers with $x^2+y^2=1$ are a point on the unit circle, hence $(\cos\vartheta,\sin\vartheta)$) — one real degree of freedom. Ask such a state two questions: (i) what are your weights? (ii) what are your weights *after* recombination? Now consider the state

$$ |{+i}\rangle = \tfrac{1}{\sqrt2}\big(|0\rangle + i\,|1\rangle\big), $$

which answers $(0.5, 0.5)$ to **both** questions. Let's check that, and then check whether *any* real state can do the same.

```python
ket_plus_i = np.array([1.0, 1.0j]) / np.sqrt(2)
print("|+i>: weights            =", np.round(np.abs(ket_plus_i) ** 2, 3))
print("|+i>: recombined weights =", np.round(np.abs(recombine(ket_plus_i)) ** 2, 3))
print()

# Every real unit qubit is (cos v, sin v). Scan v finely and look for one that is 0.5/0.5 both before AND after recombination.
v = np.linspace(0, 2 * np.pi, 20001)
real_states = np.array([np.cos(v), np.sin(v)])                       # shape (2, 20001), all real
w_before = np.abs(real_states) ** 2                                  # weights before
w_after = np.abs(recombine(real_states)) ** 2                        # weights after recombination
both_balanced = (np.abs(w_before[0] - 0.5) < 1e-3) & (np.abs(w_after[0] - 0.5) < 1e-3)
print("real states that are 0.5/0.5 before recombination :", int((np.abs(w_before[0] - 0.5) < 1e-3).sum()), " (the |+>/|-> family)")
print("real states that are 0.5/0.5 after  recombination :", int((np.abs(w_after[0] - 0.5) < 1e-3).sum()), " (the |0>/|1> family)")
print("real states that are 0.5/0.5 BOTH before and after:", int(both_balanced.sum()))
```

**Read the output.** $|{+i}\rangle$ answers $0.5/0.5$ to both questions. Among real states, the ones balanced *before* recombination are $\pm|+\rangle,\ \pm|-\rangle$, and those balanced *after* are $\pm|0\rangle,\ \pm|1\rangle$ (the counts above are the few grid points sitting within tolerance of each of those four states); **no real state is balanced under both** — that count is zero. You can see why by hand: for $(\cos\vartheta,\sin\vartheta)$ the recombined outcome-$0$ weight is $\tfrac12(\cos\vartheta+\sin\vartheta)^2 = \tfrac12(1+2\sin\vartheta\cos\vartheta) = \tfrac12(1+\sin 2\vartheta)$, equal to $\tfrac12$ only when $\sin 2\vartheta=0$, i.e. exactly at $|0\rangle$ or $|1\rangle$, which are not balanced before. So $|{+i}\rangle$ is a state that *no pair of real amplitudes can hold* — it needs a second real number inside each amplitude, and that is precisely what a complex amplitude is.

Is $|{+i}\rangle$ physical, or a mathematical curiosity? It is physical: it is the state a qubit reaches after a quarter-turn phase, and a third recombination rule (the "$Y$ basis", `01/08`–`01/09`) answers it with certainty, exactly as section 3 answered $|+\rangle$. Two honest caveats. First, a real *mixture* — a coin that secretly picked $|+\rangle$ or $|-\rangle$ — also answers $0.5/0.5$ to both questions; what separates $|{+i}\rangle$ from that mixture is the third, $Y$-type question of exercise 2, and that rule itself contains an $i$. So this is a *plausibility* argument at the level of one qubit, not a proof from inside a real theory; the definitive statement that quantum theory over the reals is experimentally distinguishable from quantum theory over $\mathbb{C}$ is a multi-party entanglement result (Renou et al., 2021 — see References). Second, we take $\mathbb{C}$ as *sufficient* — that is the first postulate of quantum theory (Nielsen & Chuang, §2.2), and the necessity direction is the one you earned today; whether anything beyond $\mathbb{C}$ could ever be needed is outside this course.

## 5. Global phase is free; relative phase is physical

One subtlety keeps everyone honest. Multiply an *entire* state by a phase $e^{i\alpha}$ — a **global phase** — and every weight is unchanged, because $|e^{i\alpha}|=1$; and since $e^{i\alpha}$ factors out of $a_0 \pm a_1$, the recombined weights are unchanged too. A global phase is physically invisible; you may add or drop one at will.

What you may *not* drop is a **relative** phase — a phase on one amplitude but not the other. That is the difference between $|+\rangle$ and $|-\rangle$, and section 3 showed it flips a certain $0$ into a certain $1$. Let's put the two side by side using only weights.

```python
alpha = np.pi / 3
plus_global = np.exp(1j * alpha) * KET_PLUS          # |+> with a global phase e^{i alpha}
for name, state in [("|+>          ", KET_PLUS), ("e^{ia}|+>    ", plus_global), ("|->          ", KET_MINUS)]:
    print(f"{name}: weights = {np.round(np.abs(state)**2, 3)}   recombined weights = {np.round(np.abs(recombine(state))**2, 3)}")
```

**Read the output.** The global phase changes nothing anyone can measure: same weights, same recombined weights $(1, 0)$ — $e^{i\alpha}|+\rangle$ *is* $|+\rangle$, physically. The relative phase that turns $|+\rangle$ into $|-\rangle$ leaves the plain weights at $0.5/0.5$ but flips the recombined weights to $(0, 1)$: as different as $|0\rangle$ and $|1\rangle$. (In `01/05` you will meet the inner product, which turns "same up to a global phase" and "as different as $|0\rangle$ and $|1\rangle$" into a single number.)
