## 2. The greatest hits — twenty promises of Modules 02 to 04, delivered

A course can claim a great deal. This one can show its work. The `course_greatest_hits()` function in `qot_course.quantum_ot.synthesis` re-runs the canonical numerical demonstrations of the course — twenty results spanning information theory, information geometry, classical transport, and the quantum lift — in a single call. Every entry is a promise some notebook made, recomputed from the validated source so you can watch it land. The list runs from the entropy of a fair coin to Amari's quantum bridge. The results of Module 01 and of the capstone (`04/12`–`04/20`) are not in it; you re-run those in their own notebooks.

The values come straight from `course_greatest_hits()`; we group them **by the module that earned each one**.


```python
# Re-run every canonical result of the course (a few seconds: this calls the real
# pipeline -- the coupling SDP prints accuracy notes, which we leave visible).
hits = course_greatest_hits()

# Each key reads "<module number>: <label>". An explicit table names the module of each
# number, and a number missing from the table stops the cell with an error: a change in the
# source cannot silently file a result under the wrong module.
MODULE_NAMES = {
    "02": "02 - Information and Geometry",
    "03": "03 - Classical Optimal Transport",
    "04": "04 - Quantum Optimal Transport",
}
MODULE_ORDER = list(MODULE_NAMES.values())


def _module_of(number: str) -> str:
    """Module that earned the result carrying this number; an unknown number is an error."""
    if number not in MODULE_NAMES:
        raise KeyError(f"greatest-hits module {number!r} has no name: add it to MODULE_NAMES")
    return MODULE_NAMES[number]


# Group the results by module, preserving the function's within-module order.
grouped: dict[str, list[tuple[str, float]]] = {m: [] for m in MODULE_ORDER}
for raw_key, value in hits.items():
    number, _, rest = raw_key.partition(":")
    grouped[_module_of(number.strip())].append((rest.strip(), value))

# Print as a module-grouped table -- the course's promises, delivered, by module.
width = max(len(lbl) for entries in grouped.values() for lbl, _ in entries)
for module in MODULE_ORDER:
    print(f"\nModule {module}")
    print("-" * (width + 18))
    for label, value in grouped[module]:
        print(f"  {label:<{width}s}  {value:>12.6f}")

print(f"\n{len(hits)} canonical results, every one recomputed live from the validated source.")
```

**Read the table.** Every line is a promise some notebook made, recomputed here from scratch — and grouped by the module that earned it.

**Module 02 — information and geometry.** The fair coin carries exactly $H = 1$ bit of surprise; mutual information sits at $0$ for independent variables and $1$ bit when they are perfectly correlated — its two endpoints. The two-geometries-on-one-simplex punchline shows in two numbers: the mixture midpoint puts almost no mass in the middle ($\approx 0$), while the Wasserstein midpoint moves the whole bump there (mass $\approx 1$). Then the quantum surprises that classical information theory forbids: the Bell pair's quantum mutual information reaches **2 bits** — twice the classical ceiling — and its conditional entropy is **$-1$ bit**, genuinely negative. And the seed of the whole course: $|+\rangle\langle+|$ and $I/2$ share a $Z$-diagonal yet differ, with relative entropy $1$ bit and a strictly positive Bures distance.

**Module 03 — classical optimal transport.** Where Monge's deterministic map fails, the Kantorovich linear program still returns a finite cost. The one-dimensional Wasserstein closed form matches the linear-program value to machine precision — the quantile gift, "sort and match". And the bridge to the quantum world is already visible: the square root of the Bures–Wasserstein matrix term on density matrices equals the Bures distance from Module 02, to the digit. Gaussians and density matrices were speaking the same language all along.

**Module 04 — quantum optimal transport.** The coupling semidefinite program delivers a strictly positive transport cost between $|+\rangle\langle+|$ and $I/2$, and the SWAP-based quantum Wasserstein squared between $|0\rangle$ and $|1\rangle$ is exactly $1$. And Amari's bridge survives the lift to operators: the regularised transport value $\mathrm{tr}(C P) - \varepsilon\, S(P)$ equals $\varepsilon$ times the Umegaki relative entropy of the plan to the matrix-exponential Gibbs kernel — the same two numbers, to rounding error.

They all hold at once. (The coupling solver prints a note that its interior-point solution "may be inaccurate"; that is the convex solver flagging its own finite tolerance on a pair with a pure marginal, whose answer sits on the boundary of the feasible set. Both coupling values of Module 04 agree with their closed forms, $0.5$ and $1$, to better than $10^{-6}$.)


## 3. What the capstone measured

One question ran through the second half of this module, and the course refused to answer it cheaply: **can a quantum-optimal-transport coupling measure read something the phase-locking value (PLV) cannot?** The phase-locking value is the workhorse of phase-synchrony analysis — the modulus of the first circular moment of the phase difference, $\mathrm{PLV} = |\langle e^{i\Delta\theta}\rangle|$. The capstone answered in steps, each one a derivation or a measurement, and the measurements could have come out otherwise.

**The redundancy reveal (`04/14`).** First, the uncomfortable truth: on the naive phase-qubit embedding, the quantum measures were PLV *in disguise*. The rank correlation $\mathrm{Spearman}(\mathrm{PLV}, \mathrm{QMI}) \approx 1$, and the operator showed exactly why — the only coupling-bearing element of $\rho_{AB}$ was an off-block coherence with $|\rho_{AB}[1,2]| = \mathrm{PLV}/4$. This was a property of the *embedding*, not of quantum optimal transport — which told us precisely what to change.

**The richer embedding (`04/15`).** Swapping the qubit for the multi-frequency qutrit $\tfrac{1}{\sqrt 3}\bigl(|0\rangle + e^{i\theta}|1\rangle + e^{2i\theta}|2\rangle\bigr)$ gave $\rho_{AB}$ a *second* coupling channel — the element $\rho_{AB}[6,2]$, carrying the **second** circular moment. Two couplings can share a PLV and differ in that second moment.

**The hand-built experiment (`04/16`).** Two ensembles were built with identical PLV by construction (first moment $a_1 = 0.4$) and a different second moment ($a_2 = 0.0$ against $0.3$). PLV could not tell them apart, and neither could the naive embedding's quantum mutual information, exactly as `04/14` predicted. The **richer** embedding's quantum mutual information separated them. So did its classical neighbour, the second-harmonic PLV $|\langle e^{2i\Delta\theta}\rangle|$, and it was reliable with 200 samples where the richer quantum measure needed more than 20,000.

**The bench (`04/17`).** Then the comparison left the hand-built ensembles for the bench of `04/17`: a pair of coupled oscillators whose second harmonic moves on its own, simulated once with its results committed. With a coupling in the second harmonic alone the exact PLV is zero, and the richer quantum measure detects the coupling. A classical statistic of the same two harmonics reads as far, and on the finite records of the bench it was reliable sooner than the richer quantum measure. The two harmonics also have a relative phase. The quantum measures computed here read its cosine only. Its sign is invisible to them, and a classical statistic, the second sine moment, detects it.

**The dial (`04/18`).** Turning down the weight of the first harmonic divided the number of samples the quantum measure needs by at least two hundred on the ensembles of `04/16`. That brought it level with the second-harmonic PLV, and no further.

**The amplitude embedding (`04/19`).** The last lever was tried on a coupling no phase statistic can see: dyads in which one amplitude follows the phase difference. PLV stayed at $0.40$ with or without the coupling. The classical statistic built for the job, the modulation index of Tort et al., adapted to a phase difference, rose from $0.000$ to $0.040$. The quantum mutual information of the amplitude–phase qubit moved too, downwards, from $0.0034$ to $0.00004$, because the phase difference gathers where the coherence of that qubit vanishes. Both detected the coupling in every set of records from 2,000 samples on; at 200 samples the modulation index did so in $0.79$ of the sets and the quantum measure in $0.32$. The change of embedding let the quantum measure respond to this coupling, and it read nothing the modulation index misses.

Put together, the accounting of the capstone is five sentences, and you can state each with its evidence.

- **A quantum coupling measure reads past PLV when the embedding carries the second harmonic.** On the naive embedding it follows PLV.
- **It reads as far as a classical statistic of the same two harmonics, and no further.**
- **On the relative phase of the two harmonics, the measures computed here read less: its cosine, and not its sign.**
- **On a phase–amplitude coupling, a change of embedding made the quantum measure respond, and it read nothing the modulation index misses.**
- **What quantum optimal transport brings is one common geometry.** The classical side is a menu of statistics, each chosen by hand. The quantum side builds one state and measures its distance from independence, by a relative entropy (quantum mutual information) or a transport distance (Bures); choosing a statistic becomes choosing an embedding. That is a unification. On the systems of the capstone it is not a gain in sensitivity.

The limits belong to the result: one hand-built pair of ensembles, one model of coupled oscillators at a handful of settings, simulated records no longer than 3,200 s, and synthetic signals throughout. The reading of the relative phase rests on one pair. "Reliable" means a detection rate of 0.8, which is a convention. The weightings of `04/18` are a few points on a dial, not an optimisation. And the amplitude result of `04/19` rests on one amplitude law, without noise, on one channel.

The result is modest, and it is solid. You asked a hard question, built comparisons that were fair to both sides, and wrote down what they showed. Knowing exactly what a tool reads, and what it does not, is what lets you use it well. That knowledge is the capstone's result, and it is yours.
